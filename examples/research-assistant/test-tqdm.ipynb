# Test tqdm Progress Bars

Test different tqdm configurations to see which works with jupyter-server-nbmodel.

```python
# Check versions
import tqdm
import ipywidgets

print(f"tqdm: {tqdm.__version__}")
print(f"ipywidgets: {ipywidgets.__version__}")

try:
    import jupyter_server_nbmodel
    print(f"jupyter_server_nbmodel: {jupyter_server_nbmodel.__version__}")
except:
    print("jupyter_server_nbmodel: not found or no __version__")
```

```python
# Test 1: tqdm.auto (what most libraries use)
import time
from tqdm.auto import tqdm

print("Test 1: tqdm.auto")
for i in tqdm(range(20)):
    time.sleep(0.2)
```

```python
# Test 2: plain tqdm
from tqdm import tqdm

print("Test 2: plain tqdm")
for i in tqdm(range(20)):
    time.sleep(0.2)
```

```python
# Test 3: tqdm.notebook
from tqdm.notebook import tqdm as tqdm_notebook

print("Test 3: tqdm.notebook")
for i in tqdm_notebook(range(20)):
    time.sleep(0.2)
```

```python
# Test 4: plain print with flush
import sys

print("Test 4: print with flush")
for i in range(20):
    print(f"Step {i+1}/20", end="\r", flush=True)
    time.sleep(0.2)
print("Done!              ")
```

```python
# Test 5: print on new lines
print("Test 5: print on new lines")
for i in range(20):
    if i % 5 == 0:
        print(f"Step {i+1}/20")
    time.sleep(0.2)
print("Done!")
```

```python
# Test 6: Simple print loop - does output appear in real-time?
import time
import sys

print("Test 6: Basic print - watch if each line appears immediately")
for i in range(10):
    print(f"[{time.strftime('%H:%M:%S')}] Step {i+1}/10")
    sys.stdout.flush()
    time.sleep(1)
print("Done!")
```

```python
# Test 7: Simulate trainer logging (uses Python logging module)
import time
import logging

# Configure like transformers does
logging.basicConfig(level=logging.INFO, format='%(message)s')
logger = logging.getLogger(__name__)

print("Test 7: Python logging (how transformers logs)")
for step in range(1, 11):
    logger.info(f"{{'loss': {2.5 - step*0.1:.4f}, 'step': {step}}}")
    time.sleep(1)
print("Done!")
```

```python
# Test 8: stderr output (some loggers use stderr)
import time
import sys

print("Test 8: stderr output")
for step in range(1, 11):
    print(f"[stderr] Step {step}/10", file=sys.stderr)
    sys.stderr.flush()
    time.sleep(1)
print("Done!")
```
