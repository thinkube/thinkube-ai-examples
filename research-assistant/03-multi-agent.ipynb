# Research Assistant: Multi-Agent System

Build a multi-agent system using CrewAI for coordinated paper analysis.

**Agents we'll build**:
- **Paper Summarizer**: Reads and summarizes new papers
- **Experiment Tracker**: Links papers to MLflow experiments
- **Insight Finder**: Discovers connections across papers

**Prerequisites**: Complete `02-langchain-rag.ipynb` first.

## Setup

Connect to platform services.

```python
import os

# Platform services
LITELLM_ENDPOINT = os.environ.get('LITELLM_ENDPOINT')
LITELLM_KEY = os.environ.get('LITELLM_MASTER_KEY')
NATS_URL = os.environ.get('NATS_URL')
MLFLOW_TRACKING_URI = os.environ.get('MLFLOW_TRACKING_URI')

print(f"LiteLLM: {LITELLM_ENDPOINT}")
print(f"NATS: {NATS_URL}")
print(f"MLflow: {MLFLOW_TRACKING_URI}")
```

---
## 1. Define Agents

Each agent has a role, goal, and backstory that guides its behavior.

```python
from crewai import Agent, Task, Crew, Process, LLM

# Configure LLM to use LiteLLM proxy with gpt-oss-20b
llm = LLM(
    model="openai/gpt-oss-20b",
    base_url=LITELLM_ENDPOINT,
    api_key=LITELLM_KEY,
    temperature=0.7,
    max_tokens=1024
)

# Paper Summarizer Agent
paper_summarizer = Agent(
    role="Research Paper Summarizer",
    goal="Extract and summarize key findings from research papers",
    backstory="""You are an expert at reading academic papers and extracting 
    the most important information. You focus on methodology, results, and 
    conclusions that would be relevant for ML practitioners.""",
    llm=llm,
    verbose=True
)

# Experiment Tracker Agent  
experiment_tracker = Agent(
    role="ML Experiment Tracker",
    goal="Connect research papers to practical ML experiments",
    backstory="""You specialize in translating research findings into actionable
    ML experiments. You identify what parameters, datasets, and techniques from
    papers could be tested in MLflow experiments.""",
    llm=llm,
    verbose=True
)

# Insight Finder Agent
insight_finder = Agent(
    role="Research Insight Analyst",
    goal="Find patterns and connections across multiple research papers",
    backstory="""You excel at synthesizing information from multiple sources.
    You identify trends, contradictions, and opportunities that emerge when
    comparing different research approaches.""",
    llm=llm,
    verbose=True
)

print("✓ Agents created: paper_summarizer, experiment_tracker, insight_finder")
```

---
## 2. Define Tasks

Tasks are specific jobs for agents to complete.

```python
# Sample paper abstract for testing
sample_paper = """
Title: LoRA: Low-Rank Adaptation of Large Language Models

Abstract: We propose Low-Rank Adaptation (LoRA), which freezes the pre-trained 
model weights and injects trainable rank decomposition matrices into each layer 
of the Transformer architecture, greatly reducing the number of trainable 
parameters for downstream tasks. Compared to GPT-3 175B fine-tuned with Adam, 
LoRA can reduce the number of trainable parameters by 10,000 times and the GPU 
memory requirement by 3 times. LoRA performs on-par or better than fine-tuning 
in model quality on RoBERTa, DeBERTa, GPT-2, and GPT-3, despite having fewer 
trainable parameters.
"""

# Task 1: Summarize the paper
summarize_task = Task(
    description=f"""Analyze this research paper and provide a summary:
    
    {sample_paper}
    
    Focus on:
    1. The main contribution
    2. Key technical approach
    3. Results and benefits""",
    expected_output="A concise summary with key findings and technical details",
    agent=paper_summarizer
)

# Task 2: Suggest experiments
experiment_task = Task(
    description="""Based on the paper summary, suggest practical ML experiments 
    that could be run to validate or extend this research.
    
    Consider:
    1. What parameters could be varied
    2. What datasets would be suitable
    3. What metrics should be tracked in MLflow""",
    expected_output="A list of 2-3 concrete experiments with parameters and metrics",
    agent=experiment_tracker
)

# Task 3: Find insights
insight_task = Task(
    description="""Synthesize the paper summary and suggested experiments to 
    identify:
    
    1. Key insights for practitioners
    2. Potential limitations or risks
    3. Future research directions""",
    expected_output="A synthesis of insights with actionable recommendations",
    agent=insight_finder
)

print("✓ Tasks created: summarize_task, experiment_task, insight_task")
```
